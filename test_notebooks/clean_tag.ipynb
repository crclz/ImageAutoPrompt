```python
# danbooru-14w.csv => danbooru.txt, 一行一个

import csv
import os

if not os.path.exists("test_notebooks"):
    os.chdir("..")

assert os.path.exists("test_notebooks")
```

```python
input_csv = "entropy/conf/tag_datasets/danbooru-14w.csv"
output_txt = "entropy/conf/tag_datasets/danbooru.txt"
```

```python
from pathlib import Path

lines = Path(input_csv).read_text("utf8").splitlines()

reader = csv.reader(lines)

tags = []

for i, item in enumerate(reader):
    original = item[0]
    space_tag = original.replace("_", " ")

    tags.append(space_tag)

tags[:10]
```

```python
txt_content = "\n".join(tags)

Path(output_txt).write_text(txt_content, "utf8")
```
